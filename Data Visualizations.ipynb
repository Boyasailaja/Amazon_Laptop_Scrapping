# Create 10 visualizations using the cleaned data.

```python
# import python libraries
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Laptop Types:1
df["TypeName"].value_counts().plot(kind="bar", color="Pink")
plt.title("Laptop Types")
plt.xlabel("Type")
plt.ylabel("Number of Laptops")
plt.show()
```

```python
# Laptops by Company: 2
sns.countplot(data=df, x="Company", color="blue")
plt.title("Laptops by Company")
plt.xticks(rotation=45)
plt.show()
```

```python
# Average Screen Size by Laptop Type: 3
sns.barplot(data=df, x="TypeName", y="Inches", color="green")
plt.title("Average Screen Size by Laptop Type")
plt.xticks(rotation=45)
plt.show()
```

```python
# Number of Laptops by Company: 4
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x="Company",
    color="blue"
)

plt.title("Number of Laptops by Company")
plt.xlabel("Company")
plt.ylabel("Number of Laptops")

plt.xticks(rotation=45)

plt.show()
```

```python
# Screen Size by Laptop Type: 5
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="TypeName",
    y="Inches",
    color="purple"
)

plt.title("Screen Size by Laptop Type")
plt.xlabel("Laptop Type")
plt.ylabel("Screen Size (Inches)")

plt.xticks(rotation=30)

plt.show()
```

```python
# RAM by Laptop Type: 6
sns.boxplot(data=df, x="TypeName", y="Ram", color="skyblue")
plt.xticks(rotation=30)
plt.title("RAM by Laptop Type")
plt.show()
```

```python
# Operating System Distribution: 7
sns.countplot(data=df, x="OpSys", color="brown")
plt.xticks(rotation=30)
plt.title("Operating System Distribution")
plt.show()
```

```python
df["Ram"] = df["Ram"].astype(str).str.extract(r"(\d+)")[0]
df["Ram"] = pd.to_numeric(df["Ram"], errors="coerce")
```

```python
df["Ram"] 
```

```python
# Average RAM by Company: 8

data = df.groupby("Company")["Ram"].mean()
data.plot(kind="bar", color="teal")
plt.title("Average RAM by Company")
plt.xlabel("Company")
plt.ylabel("Average RAM (GB)")
plt.xticks(rotation=45)
plt.show()
```

```python
# Laptop Companies Distribution using Pie Chart: 9
data = df["Company"].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(
    data,
    labels=data.index
)
plt.title("Laptop Companies Distribution")
plt.show()
```

```python
# Using Violin RAM Distribution by Laptop Type: 10

sns.violinplot(
    data=df,
    x="TypeName",
    y="Ram"
)

plt.title("RAM Distribution by Laptop Type")
plt.xlabel("Laptop Type")
plt.ylabel("RAM (GB)")

plt.xticks(rotation=30)

plt.show()
```
